```python
grouped_matches = matches.groupby("team")
```

```python
def rolling_averages(group, cols, new_cols):
    group = group.sort_values("date")
    rolling_stats = group[cols].rolling(3, closed='left').mean()
    group[new_cols] = rolling_stats
    group = group.dropna(subset=new_cols)
    return group
```

```python
cols = [ 
    'team_avg_gain',
    'team_net_yards_passing',
    'team_net_yards_rushing', 
    'team_pRate',
    'team_Y/P',
    'team_TO',
    'team_third_success_rate',
    'team_time_of_possession',
    'team_score',
    'team_first_downs',
    'team_penalties',
    'team_PenYds',
    'team_total_plays',
    'opp_avg_gain',
    'opp_net_yards_passing',
    'opp_net_yards_rushing',
    'opp_pRate',
    'opp_Y/P',
    'opp_TO',
    'opp_third_success_rate',
    'opp_time_of_possession',
    'opp_score',
    'opp_first_downs',
    'opp_penalties',
    'opp_PenYds',
    'opp_total_plays'
]
new_cols = [f"{c}_rolling" for c in cols]
```

```python
new_cols
```

```python
rolling_averages(group, cols, new_cols)
```

```python
matches_rolling = matches.groupby("team").apply(
    lambda x: rolling_averages(x, cols, new_cols)
)
```

```python
matches_rolling
```

```python
matches_rolling = matches_rolling.droplevel('team')
```

```python
matches_rolling
```

```python
matches_rolling.index = range(matches_rolling.shape[0])
```

```python
matches_rolling
```

```python
def make_predictions(data, predictors):
    train = data[data["date"] < '2013-08-01']
    test = data[data["date"] > '2013-08-01']
    rf.fit(train[predictors], train["target"])
    preds = rf.predict(test[predictors])
    combined = pd.DataFrame(dict(actual=test["target"], prediction=preds), index=test.index)
    precision = precision_score(test["target"], preds)
    return combined, precision
```

```python
combined, precision = make_predictions(matches_rolling, predictors + new_cols)
```

```python
# Get feature importances from the trained model
feature_importance = pd.DataFrame({
    'feature': predictors + new_cols,
    'importance': rf.feature_importances_
}).sort_values('importance', ascending=False)

# Display top 20 most important features
print("Top 20 Most Important Features:")
print(feature_importance.head(20))

# Create a visualization
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
plt.barh(feature_importance.head(20)['feature'], feature_importance.head(20)['importance'])
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.title('Top 20 Feature Importances in NFL Win Prediction')
plt.gca().invert_yaxis()  # Highest importance at top
plt.tight_layout()
plt.show()

# Show summary statistics
print("\nFeature Importance Statistics:")
print(feature_importance['importance'].describe())

# Show how many features account for 80% of importance
cumsum = feature_importance['importance'].cumsum()
features_for_80_percent = (cumsum <= 0.8).sum()
print(f"\nNumber of features accounting for 80% of total importance: {features_for_80_percent}")
print(f"That's {features_for_80_percent}/{len(feature_importance)} features ({100*features_for_80_percent/len(feature_importance):.1f}%)")
```
